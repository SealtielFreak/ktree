```python
import numpy as np
import pandas as pd

from ktree import TreeCluster


def generate_dataset(centers: np.ndarray, stds, n_samples=500):
    X = []
    y = []

    _, ndim = centers.shape

    for i, center in enumerate(centers):
        cluster = np.random.normal(loc=center, scale=stds[i], size=(n_samples // len(centers), ndim))
        X.append(cluster)
        y.append(np.full(n_samples // len(centers), i))

    return np.vstack(X), np.concatenate(y)
```

```python
centers = np.array([[1, 1, 1], [3, 3, 3], [2, 2, 1]])

X, y = generate_dataset(
    centers,
    stds=[.5, .5, .5],
    n_samples=100,
)


tree = TreeCluster(limit_division=2, default_distribution='mean',)

labels = tree.fit_predict(X)
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

sil = silhouette_score(X, tree.labels_)
db = davies_bouldin_score(X, tree.labels_)
ch = calinski_harabasz_score(X, tree.labels_)

print(f"Silhouette:        {sil:.3f}")
print(f"Davies-Bouldin:    {db:.3f}")
print(f"Calinski-Harabasz: {ch:.3f}")
```

```python
from sklearn.metrics import (
    adjusted_rand_score,
    normalized_mutual_info_score,
    fowlkes_mallows_score,
    homogeneity_completeness_v_measure
)

ari  = adjusted_rand_score(y, tree.labels_)
nmi  = normalized_mutual_info_score(y, tree.labels_)
fmi  = fowlkes_mallows_score(y, tree.labels_)

h, c, v = homogeneity_completeness_v_measure(y, tree.labels_)

print(f"ARI:          {ari:.3f}")
print(f"NMI:          {nmi:.3f}")
print(f"FMI:          {fmi:.3f}")
print(f"Homogeneity:  {h:.3f}  Completeness: {c:.3f}  V-measure: {v:.3f}")
```
