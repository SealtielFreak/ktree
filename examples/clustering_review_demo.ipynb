```python
import numpy as np
import pandas as pd

from ktree import TreeCluster


def generate_dataset(centers: np.ndarray, stds, n_samples=500):
    X = []
    y = []

    _, ndim = centers.shape

    for i, center in enumerate(centers):
        cluster = np.random.normal(loc=center, scale=stds[i], size=(n_samples // len(centers), ndim))
        X.append(cluster)
        y.append(np.full(n_samples // len(centers), i))

    return np.vstack(X), np.concatenate(y)
```

```python
centers = np.array([[1, 1, 1, 1, 0], [3, 3, 3, 3, 2], [2, 2, 1, 2, 3]])

data, ground_truth = generate_dataset(
    centers,
    stds=[.5, .5, .5],
    n_samples=2**16,
)


tree = TreeCluster(limit_division=2, default_distribution='mean',)

labels = tree.fit_predict(data)
```

```python
X = tree.cluster_centers_

df = pd.DataFrame(X)
df['cluster'] = tree.labels_

df
```
